```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, mean_squared_error

import seaborn as sns


```

```python
data = pd.read_csv('data/case1/test1.csv')
data['speed_krpm'] = data.speed_rpm / 1000

data.head()


```

```python



fig, axes = plt.subplots(1, 2, figsize=(8, 5))

sns.regplot(
    x='speed_krpm',
    y='power_w',
    data=data,
    ax=axes[0]
)
axes[0].set_title("Power vs Speed")

sns.regplot(
    x='flow_m3_h',
    y='power_w',
    data=data,
    ax=axes[1]
)
axes[1].set_title("Power vs Flow")

plt.tight_layout()
plt.show()


#correlation between speed and flow
print(data[['speed_krpm', 'flow_m3_h', 'power_w']].corr()) 
```
